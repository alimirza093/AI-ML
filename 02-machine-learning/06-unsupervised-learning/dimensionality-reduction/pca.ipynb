```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X, y = make_blobs(n_samples=500, centers=3, n_features=5,random_state=42,cluster_std=1.5)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(X_scaled)
x_pca
```

```python
pca_df = pd.DataFrame(
    x_pca,
    columns=['PCA1','PCA2']
)
pca_df['labels'] = y
pca_df
```

```python
sns.scatterplot(
    x=pca_df['PCA1'],
    y=pca_df['PCA2'],
    hue=pca_df['labels'],
    palette='tab10'
)
```
