```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('/home/ali-mirza/AI-ML/data/train.txt',sep=';',header=None,names=['text','emotions'])
df.head()
```

```python
emotions = df['emotions'].unique()
emotion_num = {}
i = 0
for emo in emotions:
    emotion_num[emo] = i
    i+=1
df['emotions'] = df['emotions'].map(emotion_num)
df['emotions'].unique()
```

```python
df['emotions'].head()
```

```python
#Convert all text in LowerCase

df['text'] = df['text'].apply(lambda x : x.lower())
df['text']
```

```python
#Now remove Punctuation
import string

def remove_punctuation(txt):
    return txt.translate(str.maketrans("","",string.punctuation))

df['text'] = df['text'].apply(remove_punctuation)
```

```python
#Now remove Numbers
def remove_digits(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new += i
    return new
df['text'] = df['text'].apply(remove_digits)
df['text'].head()
```

```python
#Remove links
def remove_links(txt):
    words = txt.split()

    new_words = []

    for word in words:
        if not word.startswith(("http://", "https://", "www.")):
            new_words.append(word)

    return " ".join(new_words)
df['text'] = df['text'].apply(remove_links)
df['text'].head()
```

```python
#Remove Emojis

def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new
df['text'] = df['text'].apply(remove_emojis)
df['text'].head()
```

```python
#Now remove Stop Words to reduce noise

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab")
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
def remove_stopWords(txt):
    words = word_tokenize(txt)
    cleaned = []
    for word in words:
        if not word in stop_words:
            cleaned.append(word)
    return ' '.join(cleaned)

df['text'] = df['text'].apply(remove_stopWords)
df['text'].head()
            
```
